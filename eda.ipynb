```python
import unicodedata

roca = gpd.read_file('vars_ambiental_ahp/AMB-028-AT')


def _norm(s):
    """Minusculas + sin tildes para matching robusto."""
    if s is None:
        return ''
    s = str(s).lower()
    s = unicodedata.normalize('NFKD', s)
    return ''.join(c for c in s if not unicodedata.combining(c))


# Escala de capacidad portante (1 = peor, 5 = mejor) por raiz de litologia.
# Se buscan raices (stems) sin tildes para cubrir plurales y adjetivos.
ESCALA = {
    # 5 - cristalinas competentes / plutonicas / alto grado
    'cuarcit': 5, 'gabro': 5, 'norit': 5, 'granit': 5, 'granodiorit': 5,
    'tonalit': 5, 'diorit': 5, 'monzonit': 5, 'monzodiorit': 5, 'sienit': 5,
    'aplit': 5, 'anfibolit': 5, 'granulit': 5, 'dunit': 5, 'peridotit': 5,
    'harzburg': 5, 'lerzolit': 5, 'webster': 5, 'serpentin': 5, 'picrit': 5,
    'komatiit': 5, 'rodingit': 5,
    # 4 - volcanicas competentes, gneis, arenisca, dolomia
    'basalt': 4, 'andesit': 4, 'dacit': 4, 'riolit': 4, 'riodac': 4,
    'latit': 4, 'diabas': 4, 'dolerit': 4, 'lava': 4, 'gneis': 4,
    'migmatit': 4, 'dolomit': 4, 'arenisc': 4, 'arenit': 4, 'metarenit': 4,
    'metaconglomerad': 4,
    # 3 - caliza, marmol, conglomerado, pizarra, carbonatos, chert
    'caliz': 3, 'marmol': 3, 'conglomerad': 3, 'pizarr': 3, 'bioherma': 3,
    'arrecife': 3, 'estromatol': 3, 'packstone': 3, 'chert': 3,
    # 2 - foliadas debiles, limolita, piroclasticas, marga
    'esquist': 2, 'filit': 2, 'limolit': 2, 'metalimolit': 2, 'metapelit': 2,
    'metalodolit': 2, 'marga': 2, 'toba': 2, 'tobac': 2, 'ignimbrit': 2,
    'aglomerad': 2, 'piroclast': 2, 'epiclast': 2, 'lapilli': 2, 'ceniza': 2,
    'fosforit': 2,
    # 1 - finas blandas, evaporitas, organicos, depositos no consolidados
    'lutit': 1, 'shale': 1, 'lodolit': 1, 'arcillolit': 1, 'arcilla': 1,
    'bentonit': 1, 'yeso': 1, 'halit': 1, 'diatomit': 1, 'diatome': 1,
    'carbon': 1, 'turba': 1, 'loes': 1, 'arena': 1, 'grava': 1,
    'aluvial': 1, 'coluvial': 1, 'abanico': 1, 'terraza': 1, 'glaciar': 1,
    'eolic': 1, 'duna': 1, 'paludal': 1, 'manglar': 1,
}
```

```python
import numpy as np


def puntuar(desc):
    """Devuelve el MAXIMO puntaje de las litologias detectadas en la descripcion.

    Si no se detecta ninguna litologia conocida, devuelve NaN.
    """
    texto = _norm(desc)
    encontrados = [score for raiz, score in ESCALA.items() if raiz in texto]
    return max(encontrados) if encontrados else np.nan
```

```python
roca_score = roca.copy()
roca_score['score'] = roca_score['Descripcio'].apply(puntuar)
```
